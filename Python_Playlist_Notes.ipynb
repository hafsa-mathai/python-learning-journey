    `*`

 *******`

*********`

 *******`

    *`

```python
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()


for i in range(4, 0, -1):
    for j in range(5 - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()
```

Q2. Reverse and Palindrome Checker

Take a number and reverse it using a while loop.

Then check whether the original number is a palindrome.
Input: 1221
Reversed: 1221
Palindrome: Yes

```python
num=int(input("Enter a number:"))
num=abs(num)
rev=0
while num>0:
  digit=num%10
  rev=rev*10+digit
  num=num//10
print(rev)

if num==rev:
  print("Number is palindrome")
else:
  print("nOT PALINDROme")
```

Q4. Prime Number Analyzer

Take a number from the user.
- Check whether it is prime.
- Count its divisors.
- Print all divisors.
Example:
Input: 12

Total divisors: 6
Prime: No

```python
num=int(input("Enter a number:"))
count=0
divisor=[]
i=1
while i<=num:
  if num%i==0:
    count+=1
    divisor.append(i)
  i+=1
if count==2:
  print("Prime")
else:
  print("Not Prime")
print("Divisors:",divisor)
print("Total divisors:",count)

```

Q8. Find All Prime Numbers in a Range

Take two numbers from the user.

Print all prime numbers between them.

Also calculate:
- Total number of primes.
- Sum of all prime numbers.
- Largest prime number in the range.

```python
start=int(input("Enter starting number:"))
end=int(input("Enter ending number:"))
prime_numbers=0
sum=0
largest_prime=0
i=start
while i<=end:
  j=1
  count=0
  while j<=i:
    if i%j==0:
      count+=1
    j+=1
  if count==2:
    print(i)
    prime_numbers+=1
    sum+=i
    largest_prime=i
  else:
    print("not prime")
  i+=1

print("Total number of primes:",prime_numbers)
print("Sum of all prime numbers:",sum)
print("Largest prime number in the range:",largest_prime)

```

Take an integer from the user and count how many times each digit appears.

Example:
Input: 1223334444

1 appears 1 time
2 appears 2 times
3 appears 3 times
4 appears 4 times

```python
num=int(input("enyer s nynndr:"))
digit_needed=0
while digit_needed<=9:
  count=0
  temp=num
  while temp>0:
    digit=temp%10
    if digit==digit_needed:
      count+=1
    temp//=10
  if count>0:
    print(digit_needed,"appears",count,"times")
  digit_needed+=1
```
