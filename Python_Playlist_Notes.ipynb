Q9. Prime Numbers

Ask the user to enter a number n.

Print all prime numbers between 1 and n.

Example:
Enter number: 20

2 3 5 7 11 13 17 19

```python
num=int(input("Enter a number:"))
for i in range(2,num+1):
  for j in range(2,i):
    if i%j==0:
      break
  else:
    print(i)
```

Q10. Diamond Pattern

Use nested loops to print the following pattern:

    `*`
   `***`
   
  *****`

 *******`

*********`

 *******`

  *****`

   ***`

    *`

```python
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()


for i in range(4, 0, -1):
    for j in range(5 - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()
```
