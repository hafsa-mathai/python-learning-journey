## 2. First understand the mathematical ideas

### 2.1 What is numerical data?

Numerical data is information represented by numbers. Examples include:
- the marks of students: 65, 72, 81;
- temperatures during three days: 25, 27, 24;
- distances: 2.5 km, 4.0 km, 3.2 km.

A single number is called a **scalar**. A list of numbers can be thought of as a **vector**. A rectangular arrangement of numbers is called a **matrix**.

| Mathematical idea | Example | How to imagine it |
|---|---|---|
| Scalar | 7 | One number |
| Vector | [7, 8, 9] | One row/sequence of numbers |
| Matrix | [[1, 2], [3, 4]] | Rows and columns |

We will use the word **array** for NumPy's organised collection of values. A NumPy array can represent a vector, matrix or higher-dimensional numerical data.

### 2.2 Why not use an ordinary Python list for everything?

A Python list can store many values and is very useful. NumPy arrays are specially designed for numerical work. They allow us to apply a calculation to many values at once.

For example, suppose three students have marks 50, 60 and 70. We want to add 5 bonus marks to each student. With a loop, we visit every item. With a NumPy array, we can write one calculation: `marks + 5`.

**Key idea:** vectorized operation means applying an operation to the whole array without writing the loop ourselves.

## 3. Start NumPy: import and create our first array

### Explanation before running

- `import numpy as np` loads the NumPy library. `np` is a short nickname used by almost everyone.
- `np.array(...)` converts a sequence into a NumPy array.
- The numbers inside square brackets are the elements.
- `print(...)` displays the result.

Read the code line by line before running it.

```python
import numpy as np              # Load NumPy and give it the short name np

marks_list = [50, 60, 70]       # An ordinary Python list
marks_array = np.array(marks_list)  # Convert the list into a NumPy array

print("Python list:", marks_list)   # Display the ordinary list
print("NumPy array:", marks_array)  # Display the NumPy array
print("Type:", type(marks_array))   # Check the type of the object
```

**What should you notice?** The list and array both contain 50, 60 and 70. The array is a NumPy object, so it has additional numerical abilities.

## 4. List arithmetic versus NumPy arithmetic

In ordinary mathematics, adding 5 to each number in the list [50, 60, 70] gives [55, 65, 75]. In Python, `marks_list + 5` is not a valid way to add 5 to every list item. A list's `+` operator joins lists. NumPy's `+` performs numerical addition element by element.

```python
marks = np.array([50, 60, 70])  # Create an array of three marks
bonus_marks = marks + 5         # Add 5 to every element
print("Original marks:", marks)
print("After adding 5:", bonus_marks)
```

**Line-by-line meaning:**
1. We create an array containing three marks.
2. `marks + 5` means 50+5, 60+5 and 70+5 automatically.
3–4. We display both the original and the new array.

### One more simple example: multiply every value

```python
import numpy as np
prices = np.array([100, 200, 300])  # Three prices
prices_after_discount = prices * 0.90  # Keep 90% of each price (10% discount)
print(prices_after_discount)  # Display discounted prices
```

Here, `0.90` means 90 out of 100, or 90%. Therefore, multiplying each price by 0.90 leaves 90% of the original price.

## 5. Array dimensions: 1-D, 2-D and 3-D

### 5.1 One-dimensional array (1-D)

A 1-D array is a simple sequence, such as five daily temperatures:

`[25, 27, 26, 28, 30]`

It has one direction: position 0, position 1, position 2, and so on.

### 5.2 Two-dimensional array (2-D)

A 2-D array has rows and columns. Imagine marks of two students in three tests:

\[
egin{bmatrix}70 & 75 & 80\ 60 & 65 & 72\end{bmatrix}
\]

There are 2 rows (students) and 3 columns (tests). Its shape is (2, 3).

### 5.3 Three-dimensional idea (3-D)

A 3-D array can be imagined as a stack of tables. We will only inspect a tiny example; do not worry about advanced 3-D calculations today.

```python
import numpy as np
one_d = np.array([25, 27, 26, 28, 30])  # One sequence

two_d = np.array([[70, 75, 80],        # First row: Student 1
                  [60, 65, 72]])       # Second row: Student 2

three_d = np.array([                    # Two small tables stacked together
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1-D array:", one_d)
print("1-D dimensions:", one_d.ndim)
print("2-D array:\n", two_d)
print("2-D dimensions:", two_d.ndim)
print("3-D dimensions:", three_d.ndim)
```

**Understand the output:** `ndim` tells us the number of dimensions. It does not tell us how many values are inside the array.

## 6. Shape, size and data type

These three words are different:

- `shape`: tells us the arrangement, such as 2 rows and 3 columns.
- `size`: tells us the total number of elements.
- `dtype`: tells us the data type of the elements, such as integer or decimal.

For the matrix with 2 rows and 3 columns, shape is `(2, 3)` and size is `2 × 3 = 6`.

```python
import numpy as np
student_marks = np.array([[70, 75, 80], [60, 65, 72]],dtype=np.int8)

print("Shape (rows, columns):", student_marks.shape)  # Arrangement
print("Total elements:", student_marks.size)           # Number of values
print("Data type:", student_marks.dtype)               # Type of values
```

### Small check
If an array has shape `(4, 3)`, then it has 4 rows and 3 columns. Total elements = 4 × 3 = 12.

## 7. Indexing: selecting one item

**Index** means the position number. Python starts counting positions from **0**, not 1.

| Human counting | Python index |
|---:|---:|
| First value | 0 |
| Second value | 1 |
| Third value | 2 |
| Fourth value | 3 |

For a 2-D array, we give two positions: `[row, column]`. `student_marks[0, 1]` means row 0 (first student), column 1 (second test).

```python
import numpy as np
values = np.array([10, 20, 30, 40, 50])
print("First value:", values[0])       # Index 0 is the first item
print("Third value:", values[2])       # Index 2 is the third item
print("Last value:", values[-1])       # Index -1 means the last item

marks_table = np.array([[70, 75, 80], [60, 65, 72]])
print("First student, second test:", marks_table[0, 1])
print("First student, second test:", marks_table[0][1])
print("Second student's marks:", marks_table[1, :])  # All columns in row 1
```

## 8. Slicing: selecting a part of an array

Slicing has the form `array[start:stop]`. The start position is included, but the stop position is **not included**.

For example, `values[1:4]` selects index 1, 2 and 3. It does not select index 4.

In a 2-D array we can write `array[row selection, column selection]`.

```python
import numpy as np
values = np.array([10, 20, 30, 40, 50, 60])
print(values[1:4])       # Select positions 1, 2 and 3
print(values[:3])        # From beginning up to (not including) index 3
print(values[3:])        # From index 3 to the end
print(values[::2])       # Every second value

marks_table = np.array([[70, 75, 80], [60, 65, 72], [88, 90, 95]])
print(marks_table[0:2, 1:3])  # First two rows; columns with index 1 and 2
```

**Common beginner mistake:** writing `values[1:4]` and expecting four values. Remember: the stop value is excluded, so this gives three values.

## 9. Vectorized operations: calculate without writing a loop

Suppose \(x=[1,2,3]\). Adding 10 to every element gives \([11,12,13]\). Multiplying every element by 2 gives \([2,4,6]\).

With NumPy, the same operation is written once. This is called **vectorization**. It makes code shorter and is generally efficient for numerical calculations.

```python
x = np.array([1, 2, 3, 4])

print("x + 10 =", x + 10)   # Add 10 to each element
print("x - 1 =", x - 1)     # Subtract 1 from each element
print("x * 2 =", x * 2)     # Multiply each element by 2
print("x / 2 =", x / 2)     # Divide each element by 2
print("x ** 2 =", x ** 2)   # Square each element
```

### Compare with an ordinary loop

The loop below performs the same task. Both approaches are correct; NumPy lets us write the calculation more directly.

```python
marks = np.array([45, 55, 65, 75])

# Method 1: use a loop
loop_result = []
for mark in marks:
    loop_result.append(mark + 5)

# Method 2: use one NumPy operation
array_result = marks + 5

print("Loop result:", loop_result)
print("NumPy result:", array_result)
```

## 10. Element-wise operations between two arrays

When we add two arrays of the same shape, NumPy adds matching positions. For example:

\[
[10,20,30]+[1,2,3]=[11,22,33]
\]

This is **element-wise addition**. It is not matrix multiplication.

```python
assignment1 = np.array([10, 20, 30])
assignment2 = np.array([1, 2, 3])

print("Addition:", assignment1 + assignment2)       # Match positions and add
print("Subtraction:", assignment1 - assignment2)    # Match positions and subtract
print("Multiplication:", assignment1 * assignment2) # Match positions and multiply
```

## 11. Broadcasting: a small array can work with a larger array

Broadcasting is NumPy's way of allowing arrays with compatible shapes to work together. The simplest case is adding one number to every element of an array. NumPy behaves as if that single number were available at every position.

A second simple case: add a 1-D array `[1, 2, 3]` to every row of a 2-D table that has three columns. The number of columns matches, so the operation can be performed row by row.

For now, remember this rule: **the dimensions must be compatible**. If NumPy says shapes cannot be broadcast together, check the number of rows and columns before changing the code.

```python
base_marks = np.array([[60, 70, 80],
                       [50, 65, 75]])
bonus_by_test = np.array([2, 3, 5])

updated_marks = base_marks + bonus_by_test
print("Original marks:\n", base_marks)
print("Bonus for each test:", bonus_by_test)
print("Updated marks:\n", updated_marks)
```

**Understand it:** the 3 bonus values match the 3 columns. The first bonus is added to every value in column 1, the second to column 2, and the third to column 3.

## 12. Universal functions (ufuncs): ready-made mathematical functions

A universal function applies a mathematical operation to each array element. Examples include:
- `np.sqrt(x)` — square root of each value;
- `np.round(x, 2)` — round to 2 decimal places;
- `np.abs(x)` — absolute value;
- `np.sin(x)` — sine of each value (the angle is in radians).

We will use square root and absolute value first.

```python
numbers = np.array([1, 4, 9, 16])
print("Square roots:", np.sqrt(numbers))

signed_numbers = np.array([-5, -2, 0, 3])
print("Absolute values:", np.abs(signed_numbers))

measurements = np.array([2.3456, 3.4567, 4.5678])
print("Rounded to 2 decimals:", np.round(measurements, 2))
```

## 13. Conditions and Boolean filtering

A comparison such as `marks >= 50` asks a yes/no question for every mark. NumPy returns `True` or `False` at each position. We can use these answers to select only the values that satisfy a condition.

Example: passing mark is 50. We want only marks greater than or equal to 50.

```python
marks = np.array([35, 48, 52, 67, 42, 81])

passed_condition = marks >= 50   # True for marks 50 or above
print("Pass/fail conditions:", passed_condition)

passed_marks = marks[passed_condition]  # Keep only positions where condition is True
print("Passed marks:", passed_marks)
print("Number of passed students:", len(passed_marks))
```

We can write the same filtering operation in one line: `marks[marks >= 50]`. We can also use two conditions. Use `&` for AND and place each condition inside brackets.

```python
marks = np.array([35, 48, 52, 67, 42, 81])
print("Marks from 50 to 70:", marks[(marks >= 50) & (marks <= 70)])
print("Marks below 50:", marks[marks < 50])
```

## 14. Summary statistics: mean, median, minimum, maximum and sum

These are basic mathematical ideas:

- **Sum:** add all values.
- **Mean (average):** sum divided by the number of values.
- **Median:** the middle value after arranging values from smallest to largest. For an even number of values, it is the average of the two middle values.
- **Minimum:** smallest value.
- **Maximum:** largest value.
- **Standard deviation:** a measure of how spread out values are around the mean. We will use it as a ready-made measure; detailed statistical theory is not required here.

For values 2, 4 and 6, mean = (2+4+6)/3 = 4.

```python
marks = np.array([50, 60, 70, 80, 90])

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard deviation:", np.std(marks))
```

### Summary by row or by column

For a 2-D array, `axis=0` means calculate down the rows (one answer for each column). `axis=1` means calculate across the columns (one answer for each row). Imagine a table where each row is a student and each column is a test.

```python
marks = np.array([[70, 80, 90],
                  [60, 70, 80],
                  [50, 60, 70]])

print("Average of each test (down rows):", np.mean(marks, axis=0))
print("Average of each student (across columns):", np.mean(marks, axis=1))
```

## 15. Shape and reshape

The **shape** tells us the arrangement of an array. **Reshape** changes the arrangement without changing the total number of elements.

If we have 6 values, we can arrange them as 2 rows and 3 columns or 3 rows and 2 columns. We cannot reshape 6 values into 4 rows and 2 columns because that would require 8 values.

```python
flat = np.array([1, 2, 3, 4, 5, 6])

matrix_2_by_3 = flat.reshape(2, 3)  # 2 rows and 3 columns
matrix_3_by_2 = flat.reshape(3, 2)  # 3 rows and 2 columns

print("Original:", flat)
print("2 by 3:\n", matrix_2_by_3)
print("3 by 2:\n", matrix_3_by_2)
```

## 16. Joining and splitting arrays

- `np.concatenate()` joins arrays along an existing direction.
- `np.vstack()` stacks arrays vertically (one above the other).
- `np.hstack()` places arrays side by side.
- `np.split()` divides an array into equal sections when the split is possible.

```python
first = np.array([1, 2, 3])
second = np.array([4, 5, 6])

print("Joined:", np.concatenate((first, second)))
print("Side by side:", np.hstack((first, second)))
print("Split into two:", np.split(np.array([10, 20, 30, 40]), 2))

upper = np.array([[1, 2]])
lower = np.array([[3, 4]])
print("Vertical stack:\n", np.vstack((upper, lower)))
```

## 17. Copy versus view (why changing one array can affect another)

A **copy** is a separate array. Changing it does not change the original array.

A **view** is another way of looking at the same underlying data. If the view is changed, the original data may also change. For beginners, the safe rule is: use `.copy()` when you need an independent array.

```python
original = np.array([10, 20, 30])
independent_copy = original.copy()  # Make a separate copy
independent_copy[0] = 999          # Change only the copy
print("Original after copy was changed:", original)
print("Copy:", independent_copy)

original2 = np.array([1, 2, 3, 4])
view_part = original2[1:3]          # A slice is commonly a view
view_part[0] = 200                 # Change the selected shared data
print("Original after view change:", original2)
print("View:", view_part)
```

## 18. Random numbers: create sample data

A random number is selected by chance. `np.random.default_rng(7)` creates a random-number generator. The number 7 is a **seed**: using the same seed gives the same sequence each time, which makes a teaching example reproducible.

```python
rng = np.random.default_rng(7)  # Create a reproducible random generator
random_marks = rng.integers(40, 101, size=10)  # 10 integers from 40 to 100
print("Random marks:", random_marks)
print("Average:", np.mean(random_marks))
```

## 19. A gentle introduction to vectors and matrices

A vector can represent a list of coordinates, for example \(v=[2,3]\). A matrix is a rectangular arrangement of numbers. NumPy makes matrix calculations possible.

### Matrix addition

Matrices of the same shape can be added position by position:

\[
A=egin{bmatrix}1&2\3&4\end{bmatrix},\quad
B=egin{bmatrix}10&20\30&40\end{bmatrix}
\]

\[
A+B=egin{bmatrix}11&22\33&44\end{bmatrix}
\]

### Matrix multiplication is different from element-wise multiplication

- `A * B` multiplies matching positions.
- `A @ B` performs matrix multiplication.

For matrix multiplication, the number of columns in A must equal the number of rows in B. A (2×2) matrix can multiply a (2×2) matrix. We will only try a small example.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])

print("Element-wise multiplication A * B:\n", A * B)
print("Matrix multiplication A @ B:\n", A @ B)
print("Transpose of A:\n", A.T)
```

**Optional mathematical extension:** The dot product of \([1,2]\) and \([3,4]\) is \(1×3+2×4=11\). NumPy calculates it with `np.dot`.

```python
vector_a = np.array([1, 2])
vector_b = np.array([3, 4])
print("Dot product:", np.dot(vector_a, vector_b))
```

## 20. Mini application: temperature measurements

Suppose we record the temperatures for five days. We want to:
1. store the values in an array;
2. convert Celsius to Fahrenheit using \(F=(9/5)C+32\);
3. find the average, highest and lowest temperature;
4. find days with temperature above 28°C.

This example connects a mathematical formula to an array operation.

```python
days_temperature_c = np.array([25.0, 27.5, 29.0, 26.0, 30.5])

temperature_f = (9 / 5) * days_temperature_c + 32  # Apply formula to every value

print("Celsius:", days_temperature_c)
print("Fahrenheit:", np.round(temperature_f, 1))
print("Average Celsius:", np.mean(days_temperature_c))
print("Highest Celsius:", np.max(days_temperature_c))
print("Lowest Celsius:", np.min(days_temperature_c))
print("Days above 28 C:", days_temperature_c[days_temperature_c > 28])
```
